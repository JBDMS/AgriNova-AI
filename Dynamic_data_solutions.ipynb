```python
import hashlib

def file_hash(path):

    h = hashlib.md5()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(8192)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()


hashes = []

for path in tqdm(image_df["filepath"]):
    try:
        hashes.append(file_hash(path))
    except:
        hashes.append(None)

image_df["file_hash"] = hashes

print(
    "Duplicate files:",
    image_df["file_hash"].duplicated().sum()
)
```

```python
IMG_SIZE = (224, 224)

def load_rgb(path):

    img = cv2.imread(str(path))

    if img is None:
        return None

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    return img


def preprocess_image(path):

    img = load_rgb(path)

    if img is None:
        return None

    img = cv2.resize(
        img,
        IMG_SIZE,
        interpolation=cv2.INTER_AREA
    )

    return img
```

```python
def extract_rgb_features(img):

    img_float = img.astype(np.float32) / 255.0

    features = {}

    channels = {
        "R": img_float[:, :, 0],
        "G": img_float[:, :, 1],
        "B": img_float[:, :, 2]
    }

    for name, channel in channels.items():

        features[f"{name}_mean"] = channel.mean()
        features[f"{name}_std"] = channel.std()
        features[f"{name}_min"] = channel.min()
        features[f"{name}_max"] = channel.max()

    return features
```

```python
def extract_hsv_features(img):

    hsv = cv2.cvtColor(
        img,
        cv2.COLOR_RGB2HSV
    ).astype(np.float32)

    hsv[:, :, 0] /= 179.0
    hsv[:, :, 1] /= 255.0
    hsv[:, :, 2] /= 255.0

    names = ["H", "S", "V"]

    features = {}

    for i, name in enumerate(names):

        channel = hsv[:, :, i]

        features[f"{name}_mean"] = channel.mean()
        features[f"{name}_std"] = channel.std()
        features[f"{name}_min"] = channel.min()
        features[f"{name}_max"] = channel.max()

    return features
```

```python
def extract_lab_features(img):

    lab = cv2.cvtColor(
        img,
        cv2.COLOR_RGB2LAB
    ).astype(np.float32)

    features = {}

    names = ["L", "A", "B"]

    for i, name in enumerate(names):

        channel = lab[:, :, i]

        features[f"LAB_{name}_mean"] = channel.mean()
        features[f"LAB_{name}_std"] = channel.std()
        features[f"LAB_{name}_min"] = channel.min()
        features[f"LAB_{name}_max"] = channel.max()

    return features
```

```python
def extract_vegetation_features(img):

    img_float = img.astype(np.float32) + 1e-6

    R = img_float[:, :, 0]
    G = img_float[:, :, 1]
    B = img_float[:, :, 2]

    features = {}

    # Green Excess Index
    ExG = (2 * G) - R - B

    # Green-Red ratio
    GR = G / (R + 1e-6)

    # Normalized Green Red Difference
    NGRDI = (G - R) / (G + R + 1e-6)

    indices = {
        "ExG": ExG,
        "G_R_ratio": GR,
        "NGRDI": NGRDI
    }

    for name, value in indices.items():

        features[f"{name}_mean"] = np.mean(value)
        features[f"{name}_std"] = np.std(value)
        features[f"{name}_min"] = np.min(value)
        features[f"{name}_max"] = np.max(value)

    return features
```

```python
def extract_texture_features(img):

    gray = cv2.cvtColor(
        img,
        cv2.COLOR_RGB2GRAY
    )

    # Reduce image to 8-bit levels
    gray_small = cv2.resize(
        gray,
        (128, 128)
    )

    gray_quantized = (
        gray_small / 32
    ).astype(np.uint8)

    glcm = graycomatrix(
        gray_quantized,
        distances=[1],
        angles=[0],
        levels=8,
        symmetric=True,
        normed=True
    )

    features = {}

    properties = [
        "contrast",
        "dissimilarity",
        "homogeneity",
        "energy",
        "correlation"
    ]

    for prop in properties:

        features[
            f"texture_{prop}"
        ] = graycoprops(
            glcm,
            prop
        )[0, 0]

    return features
```

```python
# ============================================================
# CELL 17: EXTRACT IMAGE FEATURES
# ============================================================

# If original_df is empty, use all validated images
if len(original_df) == 0:
    original_df = image_df[
        image_df["valid"] == True
    ].copy()

print("Images used for feature extraction:", len(original_df))


# ------------------------------------------------------------
# Extract all features from one image
# ------------------------------------------------------------

def extract_all_features(path):

    img = preprocess_image(path)

    if img is None:
        return None

    features = {}

    # RGB features
    features.update(
        extract_rgb_features(img)
    )

    # HSV features
    features.update(
        extract_hsv_features(img)
    )

    # LAB features
    features.update(
        extract_lab_features(img)
    )

    # Vegetation-index features
    features.update(
        extract_vegetation_features(img)
    )

    # Texture features
    features.update(
        extract_texture_features(img)
    )

    return features


# ------------------------------------------------------------
# Extract features for all images
# ------------------------------------------------------------

feature_records = []

for _, row in tqdm(
    original_df.iterrows(),
    total=len(original_df),
    desc="Extracting features"
):

    try:

        features = extract_all_features(
            row["filepath"]
        )

        if features is not None:

            # Keep image information
            features["filepath"] = row["filepath"]
            features["class"] = row["class"]

            feature_records.append(features)

    except Exception as e:

        print(
            "Error:",
            row["filepath"],
            "|",
            e
        )


# ------------------------------------------------------------
# Create feature dataframe
# ------------------------------------------------------------

feature_df = pd.DataFrame(feature_records)


# ------------------------------------------------------------
# Basic validation
# ------------------------------------------------------------

print("\n========================================")
print("FEATURE EXTRACTION COMPLETE")
print("========================================")

print(
    "Feature dataset shape:",
    feature_df.shape
)

print(
    "Number of images:",
    len(feature_df)
)

print(
    "Number of features:",
    len(feature_df.columns)
)

print("\nClass distribution:")
print(
    feature_df["class"].value_counts()
)

print("\nMissing values:")
print(
    feature_df.isnull().sum().sum()
)


# ------------------------------------------------------------
# Display first 5 rows
# ------------------------------------------------------------

feature_df.head()
```

```python
def calculate_metrics(y_true, y_pred):

    return {
        "Accuracy": accuracy_score(
            y_true,
            y_pred
        ),

        "Precision_macro": precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "Recall_macro": recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        ),

        "F1_macro": f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    }


results = []

for name, pred in [
    ("Logistic Regression", lr_pred),
    ("SVM", svm_pred),
    ("Random Forest", rf_pred)
]:

    metrics = calculate_metrics(
        y_test,
        pred
    )

    metrics["Model"] = name

    results.append(metrics)


model_results = pd.DataFrame(results)

model_results = model_results[
    [
        "Model",
        "Accuracy",
        "Precision_macro",
        "Recall_macro",
        "F1_macro"
    ]
]

model_results
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

CNN_SIZE = (224, 224)
BATCH_SIZE = 32
```

```python
cnn_df = original_df[
    ["filepath", "class"]
].copy()

cnn_df["class"] = cnn_df["class"].astype(str)

train_df, temp_df = train_test_split(
    cnn_df,
    test_size=0.30,
    stratify=cnn_df["class"],
    random_state=SEED
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["class"],
    random_state=SEED
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.10,
    height_shift_range=0.10,
    zoom_range=0.10,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)

train_gen = train_datagen.flow_from_dataframe(
    train_df,
    x_col="filepath",
    y_col="class",
    target_size=CNN_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=True,
    seed=SEED
)

val_gen = test_datagen.flow_from_dataframe(
    val_df,
    x_col="filepath",
    y_col="class",
    target_size=CNN_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

test_gen = test_datagen.flow_from_dataframe(
    test_df,
    x_col="filepath",
    y_col="class",
    target_size=CNN_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)
```

```python
num_classes = len(train_gen.class_indices)

base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.trainable = False

model = keras.Sequential([
    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.30),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dropout(0.20),

    layers.Dense(
        num_classes,
        activation="softmax"
    )
])

model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=1e-3
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)

model.summary()
```

```python
callbacks = [

    EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=1e-6
    )
]


history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=25,
    callbacks=callbacks
)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("CNN Training and Validation Accuracy")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("CNN Training and Validation Loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def make_gradcam_heatmap(
    img_array,
    model,
    last_conv_layer_name
):

    grad_model = tf.keras.models.Model(
        [
            model.inputs
        ],
        [
            model.get_layer(
                last_conv_layer_name
            ).output,
            model.output
        ]
    )

    with tf.GradientTape() as tape:

        conv_outputs, predictions = grad_model(
            img_array
        )

        pred_index = tf.argmax(
            predictions[0]
        )

        class_channel = predictions[
            :, pred_index
        ]

    grads = tape.gradient(
        class_channel,
        conv_outputs
    )

    pooled_grads = tf.reduce_mean(
        grads,
        axis=(0, 1, 2)
    )

    conv_outputs = conv_outputs[0]

    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]

    heatmap = tf.squeeze(heatmap)

    heatmap = tf.maximum(
        heatmap,
        0
    )

    heatmap /= (
        tf.math.reduce_max(heatmap)
        + tf.keras.backend.epsilon()
    )

    return heatmap.numpy()
```

```python
from sklearn.utils import resample

def bootstrap_accuracy(
    y_true,
    y_pred,
    n_bootstrap=1000
):

    rng = np.random.default_rng(SEED)

    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    scores = []

    for _ in range(n_bootstrap):

        indices = rng.integers(
            0,
            len(y_true),
            len(y_true)
        )

        scores.append(
            accuracy_score(
                y_true[indices],
                y_pred[indices]
            )
        )

    lower = np.percentile(
        scores,
        2.5
    )

    upper = np.percentile(
        scores,
        97.5
    )

    return (
        np.mean(scores),
        lower,
        upper
    )


mean_acc, lower, upper = bootstrap_accuracy(
    cnn_true,
    cnn_pred
)

print(
    f"Accuracy: {mean_acc:.4f}"
)

print(
    f"95% CI: [{lower:.4f}, {upper:.4f}]"
)
```

```python
# ============================================================
# IMAGE-BASED TRIAGE
# ============================================================

def predict_image_triage(
    image_path,
    cnn_model,
    class_names,
    confidence_threshold=IMAGE_CONFIDENCE_THRESHOLD
):

    """
    Returns:
        predicted_class
        confidence
        triage_decision
        advisory_type
    """

    img = cv2.imread(str(image_path))

    if img is None:
        return {
            "predicted_class": "unreadable",
            "confidence": 0.0,
            "triage_decision": "soil_tissue_testing",
            "advisory_type": "manual_field_inspection"
        }

    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (224, 224))
    img_array = np.expand_dims(img.astype(np.float32) / 255.0, axis=0)

    probabilities = cnn_model.predict(
        img_array,
        verbose=0
    )[0]

    predicted_index = int(np.argmax(probabilities))
    confidence = float(np.max(probabilities))

    predicted_class = class_names[predicted_index]

    if confidence >= confidence_threshold:
        triage_decision = "disease_management"
        advisory_type = f"possible_{predicted_class}"
    else:
        triage_decision = "soil_tissue_testing"
        advisory_type = "uncertain_symptom_field_sampling"

    return {
        "predicted_class": predicted_class,
        "confidence": confidence,
        "triage_decision": triage_decision,
        "advisory_type": advisory_type
    }
```

```python
# ============================================================
# STANDARD STCR TARGETED-YIELD EQUATION
# ============================================================

def stcr_fertilizer_dose(
    nutrient_requirement_NR,
    soil_contribution_CS,
    fertilizer_contribution_CF,
    target_yield_T,
    soil_test_value_SN,
    organic_manure_nutrient_ON=0.0,
    organic_coefficient_c=0.0
):

    """
    Standard STCR targeted-yield fertilizer equation:

    FD = (NR / %CF) * 100 * T - (%CS / %CF) * SN

    Optional organic-manure adjustment:

    FD = FD - c * ON

    Negative doses are set to zero because the soil already
    supplies enough nutrient for the selected target yield.
    """

    if fertilizer_contribution_CF <= 0:
        raise ValueError(
            "Fertilizer contribution %CF must be greater than zero."
        )

    slope_a = (
        nutrient_requirement_NR /
        fertilizer_contribution_CF
    ) * 100

    slope_b = (
        soil_contribution_CS /
        fertilizer_contribution_CF
    )

    raw_dose = (
        slope_a * target_yield_T
        - slope_b * soil_test_value_SN
        - organic_coefficient_c * organic_manure_nutrient_ON
    )

    recommended_dose = max(0.0, raw_dose)

    return {
        "STCR_slope_a": slope_a,
        "STCR_slope_b": slope_b,
        "raw_STCR_dose": raw_dose,
        "recommended_STCR_dose": recommended_dose
    }
```

```python
# ============================================================
# CELL 3: CREATE stcr_utils.py
# ============================================================

stcr_code = '''
import numpy as np


def stcr_fertilizer_dose(
    nutrient_requirement_NR,
    soil_contribution_CS,
    fertilizer_contribution_CF,
    target_yield_T,
    soil_test_value_SN,
    organic_manure_nutrient_ON=0.0,
    organic_coefficient_c=0.0
):
    """
    Standard STCR targeted-yield fertilizer equation:

    FD = (NR / %CF) * 100 * T - (%CS / %CF) * SN

    Negative doses are set to zero.
    """

    if fertilizer_contribution_CF <= 0:
        raise ValueError("Fertilizer contribution %CF must be greater than zero.")

    slope_a = (nutrient_requirement_NR / fertilizer_contribution_CF) * 100
    slope_b = soil_contribution_CS / fertilizer_contribution_CF

    raw_dose = (
        slope_a * target_yield_T
        - slope_b * soil_test_value_SN
        - organic_coefficient_c * organic_manure_nutrient_ON
    )

    recommended_dose = max(0.0, float(raw_dose))

    return {
        "slope_a": slope_a,
        "slope_b": slope_b,
        "raw_dose": raw_dose,
        "recommended_dose": recommended_dose
    }


def calculate_stcr_recommendation(
    target_yield_q_ha,
    soil_N_kg_ha,
    soil_P_kg_ha,
    soil_K_kg_ha,
    organic_manure_N_kg_ha=0.0,
    organic_coefficient_N=0.0,
    use_local_calibration=False,
    local_parameters=None
):
    """
    Calculates STCR-based N, P and K fertilizer recommendations.

    Default parameters are demonstration placeholders.
    Replace them with locally calibrated Uttar Pradesh STCR values
    before real field use.
    """

    if use_local_calibration and local_parameters is not None:
        parameters = local_parameters
    else:
        parameters = {
            "N": {"NR": 1.80, "CS": 35.0, "CF": 45.0},
            "P": {"NR": 0.60, "CS": 40.0, "CF": 30.0},
            "K": {"NR": 1.50, "CS": 25.0, "CF": 60.0}
        }

    n_result = stcr_fertilizer_dose(
        nutrient_requirement_NR=parameters["N"]["NR"],
        soil_contribution_CS=parameters["N"]["CS"],
        fertilizer_contribution_CF=parameters["N"]["CF"],
        target_yield_T=target_yield_q_ha,
        soil_test_value_SN=soil_N_kg_ha,
        organic_manure_nutrient_ON=organic_manure_N_kg_ha,
        organic_coefficient_c=organic_coefficient_N
    )

    p_result = stcr_fertilizer_dose(
        nutrient_requirement_NR=parameters["P"]["NR"],
        soil_contribution_CS=parameters["P"]["CS"],
        fertilizer_contribution_CF=parameters["P"]["CF"],
        target_yield_T=target_yield_q_ha,
        soil_test_value_SN=soil_P_kg_ha
    )

    k_result = stcr_fertilizer_dose(
        nutrient_requirement_NR=parameters["K"]["NR"],
        soil_contribution_CS=parameters["K"]["CS"],
        fertilizer_contribution_CF=parameters["K"]["CF"],
        target_yield_T=target_yield_q_ha,
        soil_test_value_SN=soil_K_kg_ha
    )

    return {
        "N": n_result,
        "P": p_result,
        "K": k_result,
        "recommended_N_kg_ha": n_result["recommended_dose"],
        "recommended_P_kg_ha": p_result["recommended_dose"],
        "recommended_K_kg_ha": k_result["recommended_dose"]
    }


def interpret_stcr_result(stcr_result, farmer_N_kg_ha=None):
    """
    Creates a simple interpretation of the STCR recommendation.
    """

    recommended_N = stcr_result["recommended_N_kg_ha"]
    recommended_P = stcr_result["recommended_P_kg_ha"]
    recommended_K = stcr_result["recommended_K_kg_ha"]

    interpretation = []

    interpretation.append(
        f"For the selected target yield, the STCR equation recommends "
        f"{recommended_N:.1f} kg N/ha, {recommended_P:.1f} kg P/ha and "
        f"{recommended_K:.1f} kg K/ha."
    )

    if farmer_N_kg_ha is not None and farmer_N_kg_ha > 0:

        difference = farmer_N_kg_ha - recommended_N

        if difference > 0:
            interpretation.append(
                f"The farmer-practice nitrogen dose is {difference:.1f} kg/ha "
                "higher than the STCR recommendation, indicating possible "
                "over-application for the selected yield target."
            )
        elif difference < 0:
            interpretation.append(
                f"The farmer-practice nitrogen dose is {abs(difference):.1f} kg/ha "
                "lower than the STCR recommendation, indicating a possible "
                "nutrient gap for the selected yield target."
            )
        else:
            interpretation.append(
                "The farmer-practice nitrogen dose matches the STCR recommendation."
            )

    if recommended_N == 0:
        interpretation.append(
            "The soil-test nitrogen level is sufficient for the selected "
            "yield target; no additional nitrogen is recommended."
        )

    return " ".join(interpretation)
'''

with open(PROJECT_DIR / "stcr_utils.py", "w", encoding="utf-8") as f:
    f.write(stcr_code)

print("Created:", PROJECT_DIR / "stcr_utils.py")
```

```python
# ============================================================
# CELL 4: CREATE model_utils.py
# ============================================================

model_utils_code = '''
import numpy as np
import cv2
import tensorflow as tf


def load_rice_model(model_path):
    """
    Loads the trained MobileNetV2 rice model.
    """

    return tf.keras.models.load_model(model_path)


def preprocess_leaf_image(uploaded_file, image_size=(224, 224)):
    """
    Converts an uploaded image into MobileNetV2 input format.
    """

    file_bytes = uploaded_file.read()
    image_array = np.frombuffer(file_bytes, np.uint8)

    image = cv2.imdecode(image_array, cv2.IMREAD_COLOR)

    if image is None:
        raise ValueError("Uploaded file could not be read as an image.")

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, image_size, interpolation=cv2.INTER_AREA)
    image = image.astype("float32") / 255.0
    image = np.expand_dims(image, axis=0)

    return image


def predict_leaf_condition(model, image_array, class_names):
    """
    Returns predicted class, confidence and class probabilities.
    """

    probabilities = model.predict(image_array, verbose=0)[0]
    predicted_index = int(np.argmax(probabilities))
    confidence = float(np.max(probabilities))

    return {
        "predicted_class": class_names[predicted_index],
        "confidence": confidence,
        "probabilities": {
            class_names[i]: float(probabilities[i])
            for i in range(len(class_names))
        }
    }


def apply_confidence_gate(prediction, high_confidence_threshold=0.80):
    """
    Applies the triage rule:

    High confidence -> disease management
    Low confidence -> soil/tissue testing

    The model does not diagnose Zn, B, Fe or Mn deficiency.
    """

    predicted_class = prediction["predicted_class"]
    confidence = prediction["confidence"]

    if confidence >= high_confidence_threshold:
        triage_decision = "disease_management"
        advisory = (
            f"High-confidence prediction: {predicted_class}. "
            "Confirm with an agronomist or extension officer and follow "
            "recommended disease-management practices."
        )
    else:
        triage_decision = "soil_tissue_testing"
        advisory = (
            "Model confidence is insufficient for a reliable disease decision. "
            "Collect representative leaf, soil and plant-tissue samples for "
            "laboratory diagnosis."
        )

    return {
        "triage_decision": triage_decision,
        "advisory": advisory
    }
'''

with open(PROJECT_DIR / "model_utils.py", "w", encoding="utf-8") as f:
    f.write(model_utils_code)

print("Created:", PROJECT_DIR / "model_utils.py")
```

```python
# ============================================================
# CELL 3: IMAGE TRIAGE + STCR RECOMMENDATION
# ============================================================

def stcr_dose(NR, CS, CF, target_yield, soil_test_value):
    dose = (NR / CF) * 100 * target_yield - (CS / CF) * soil_test_value
    return max(0.0, dose)


def analyze_field(change):

    with output:

        clear_output()

        if not uploader.value:
            print("Please upload a rice leaf image first.")
            return

        uploaded_file = list(uploader.value.values())[0]
        file_name = uploaded_file["metadata"]["name"]
        file_bytes = uploaded_file["content"]

        image_path = UPLOAD_DIR / file_name

        with open(image_path, "wb") as f:
            f.write(file_bytes)

        print("=" * 60)
        print("RICE CROP HEALTH RECOMMENDATION")
        print("=" * 60)
        print(f"Field ID: {field_id.value}")
        print(f"District: {district.value}")
        print()

        # ---------------- IMAGE TRIAGE ----------------

        img = cv2.imread(str(image_path))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (224, 224))
        img_array = np.expand_dims(img.astype("float32") / 255.0, axis=0)

        probabilities = model.predict(img_array, verbose=0)[0]
        predicted_index = int(np.argmax(probabilities))
        confidence = float(np.max(probabilities))

        class_names = list(train_gen.class_indices.keys())
        predicted_class = class_names[predicted_index]

        print("LEAF IMAGE TRIAGE")
        print("-" * 60)
        print(f"Predicted condition: {predicted_class}")
        print(f"Confidence: {confidence * 100:.1f}%")
        print()

        if confidence >= confidence_threshold.value:
            triage_decision = "Disease management advisory"
            advice = (
                f"High-confidence prediction: {predicted_class}. "
                "Confirm with an agronomist and follow recommended "
                "disease-management practices."
            )
        else:
            triage_decision = "Soil and tissue testing recommended"
            advice = (
                "Model confidence is low. Collect soil and plant-tissue "
                "samples for laboratory diagnosis."
            )

        print(f"Triage decision: {triage_decision}")
        print(advice)
        print()

        print("Class probabilities:")
        for name, prob in zip(class_names, probabilities):
            print(f"- {name}: {prob * 100:.1f}%")

        plt.figure(figsize=(7, 3))
        plt.bar(class_names, probabilities)
        plt.title("Leaf Image Prediction Probabilities")
        plt.ylabel("Probability")
        plt.xticks(rotation=20)
        plt.tight_layout()
        plt.show()

        # ---------------- STCR ADVISORY ----------------

        N = stcr_dose(1.80, 35, 45, target_yield.value, soil_N.value)
        P = stcr_dose(0.60, 40, 30, target_yield.value, soil_P.value)
        K = stcr_dose(1.50, 25, 60, target_yield.value, soil_K.value)

        print()
        print("STCR FERTILIZER RECOMMENDATION")
        print("-" * 60)
        print(f"Target yield: {target_yield.value:.0f} q/ha")
        print(f"Recommended N: {N:.1f} kg/ha")
        print(f"Recommended P: {P:.1f} kg/ha")
        print(f"Recommended K: {K:.1f} kg/ha")
        print()

        difference = farmer_N.value - N

        if difference > 0:
            print(
                f"Farmer N dose is {difference:.1f} kg/ha higher than "
                "the STCR recommendation."
            )
        elif difference < 0:
            print(
                f"Farmer N dose is {abs(difference):.1f} kg/ha lower than "
                "the STCR recommendation."
            )
        else:
            print("Farmer N dose matches the STCR recommendation.")

        print()
        print("Note: Leaf images do not diagnose Zn, B, Fe or Mn deficiency.")
        print("Use soil and tissue tests for nutrient confirmation.")
        print("Use locally calibrated UP STCR parameters for real advice.")

        # ---------------- SAVE RESULT ----------------

        result = {
            "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "field_id": field_id.value,
            "district": district.value,
            "image_path": str(image_path),
            "predicted_class": predicted_class,
            "confidence": confidence,
            "triage_decision": triage_decision,
            "target_yield_q_ha": target_yield.value,
            "soil_N_kg_ha": soil_N.value,
            "soil_P_kg_ha": soil_P.value,
            "soil_K_kg_ha": soil_K.value,
            "farmer_N_kg_ha": farmer_N.value,
            "STCR_N_kg_ha": N,
            "STCR_P_kg_ha": P,
            "STCR_K_kg_ha": K
        }

        pd.DataFrame([result]).to_csv(
            DASHBOARD_DIR / "dashboard_recommendations.csv",
            mode="a",
            header=not (DASHBOARD_DIR / "dashboard_recommendations.csv").exists(),
            index=False
        )

        print()
        print("Result saved to:")
        print(DASHBOARD_DIR / "dashboard_recommendations.csv")


analyze_button.on_click(analyze_field)
```
