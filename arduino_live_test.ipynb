```python
import serial
import pandas as pd
from IPython.display import display, clear_output
import time

ser = serial.Serial("COM3", 115200, timeout=3)
time.sleep(2)
ser.reset_input_buffer()

columns = [
    "Soil Moisture (Raw)",
    "Air Temperature (°C)",
    "Humidity (%)",
    "Light (Lux)",
    "Soil Temperature (°C)"
]

try:
    while True:
        line = ser.readline().decode(
            "utf-8", errors="ignore"
        ).strip()

        values = line.split(",")

        if len(values) != 5:
            continue

        try:
            readings = [float(v) for v in values]
        except ValueError:
            continue

        clear_output(wait=True)
        print("🌾 AGRINOVA AI — LIVE FIELD SENSOR DASHBOARD")
        display(pd.DataFrame({
            "Sensor": columns,
            "Live Reading": readings
        }))
        time.sleep(0.1)

except KeyboardInterrupt:
    print("Monitoring stopped.")

finally:
    ser.close()
```
